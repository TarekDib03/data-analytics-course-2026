```python
query = """
SELECT department, 
       employee_id,
       first_name || ' ' || last_name AS employee_name, 
       salary, 
       AVG(salary) OVER (PARTITION BY department) AS avg_department_salary
FROM employees
ORDER BY department DESC
LIMIT 10;
"""
pd.read_sql_query(query, conn)    # Display the average salary per department using window functions (partition by department)
```

```python
query = """
SELECT department,
       employee_name,
       salary
FROM (
    SELECT department,
           first_name || ' ' || last_name AS employee_name,
           salary,
           -- Create a separate ranking within each department. Now we can use the ranking to filter out the top earners in each department 
           -- since we have a ranking for each employee within their department inside the subquery (salary_rank table).
           RANK() OVER (
               PARTITION BY department
               ORDER BY salary DESC
           ) AS salary_rank
    FROM employees
) AS ranked_employees
WHERE salary_rank = 1;
"""
pd.read_sql_query(query, conn) 
```

```python
# 2nd way using FIRST_VALUE() Window Function and Common Table Expression (CTE)
query = """
WITH max_salary_per_department AS (
    SELECT department,
           first_name || ' ' || last_name AS employee_name,
           salary,
           FIRST_VALUE(salary) OVER(
                    PARTITION BY department 
                    ORDER BY salary DESC) AS highest_employee_salary_per_department
FROM employees
)

SELECT employee_name, 
       department, 
       salary
FROM max_salary_per_department
WHERE salary = highest_employee_salary_per_department
GROUP BY employee_name, department
ORDER BY salary DESC;
"""

pd.read_sql_query(query, conn)
```

```python
query = """
-- Create a common table expression to retrieve the customer_id, customer_name, segment and how many 
-- times the customer has purchased from the mall 

WITH customer_purchase AS (
	SELECT s.customer_id, 
           c.customer_name, 
           c.segment,
	       COUNT(*) AS purchase_count
	FROM sales s
	JOIN customers c
	ON s.customer_id = c.customer_id
	GROUP BY s.customer_id, c.customer_name, c.segment
	ORDER BY s.customer_id)

-- Number each customer by how many purchases they've made

SELECT *,
	   ROW_NUMBER() OVER (
    				ORDER BY purchase_count DESC
                    ) AS Row_N
FROM customer_purchase
ORDER BY purchase_count DESC;    
"""
pd.read_sql_query(query, conn)
```

```python
query = """
-- Create a common table expression to retrieve the customer_id, customer_name, segment and how many 
-- times the customer has purchased from the mall 

WITH customer_purchase AS (
	SELECT s.customer_id, 
           c.customer_name, 
           c.segment,
	       COUNT(*) AS purchase_count
	FROM sales s
	JOIN customers c
	ON s.customer_id = c.customer_id
	GROUP BY s.customer_id, c.customer_name, c.segment
	ORDER BY s.customer_id)

-- Number each customer by their customer segment
-- and by how many purchases they've made in descending order

SELECT *,
	   ROW_NUMBER() OVER (
    				PARTITION BY segment
				   	ORDER BY purchase_count DESC
                    ) AS Row_N
FROM customer_purchase
ORDER BY segment, purchase_count DESC;    
"""
pd.read_sql_query(query, conn)
```

```python
query = """
-- Retrieve the first_name, last_name, department, and hire_date of all employees. 
-- Add a new column called first_emp_date that returns the hire date of the first hired employee.
-- Find the difference between the hire date of the first employee hired and every other employees

SELECT first_name,
       last_name,
       department,
       hire_date,
       FIRST_VALUE(hire_date) OVER(ORDER BY hire_date) AS first_emp_date,
       CAST(julianday(hire_date) - julianday(FIRST_VALUE(hire_date) OVER(ORDER BY hire_date)) AS INT) AS days_difference
FROM employees;
"""

pd.read_sql_query(query, conn)
```

```python
query = """
-- Exercise 7.2: Find the difference between the hire date of the 
-- first employee hired and every other employees partitioned by department

SELECT *, CAST(julianday(a.hire_date) - julianday(a.first_emp_date) AS INT) AS days_diff
FROM (
    SELECT first_name, 
            last_name, 
            department, 
            hire_date,
            FIRST_VALUE(hire_date) OVER (
                PARTITION BY department
			    ORDER BY hire_date) AS first_emp_date
    FROM employees) a
ORDER BY department;
"""

pd.read_sql_query(query, conn)
```

```python
# To find the number of months between the 2 dates

query = """
SELECT *, (strftime('%Y', a.hire_date) - strftime('%Y', a.first_emp_date)) * 12 + 
                        (strftime('%m', a.hire_date) - strftime('%m', a.first_emp_date)) AS months_diff
FROM (
    SELECT first_name, 
            last_name, 
            department, 
            hire_date,
            FIRST_VALUE(hire_date) OVER (
                PARTITION BY department
			    ORDER BY hire_date) AS first_emp_date
    FROM employees) a
ORDER BY department;
"""

pd.read_sql_query(query, conn)
```
