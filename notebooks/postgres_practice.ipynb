```python
import os
import pandas as pd
from urllib.parse import quote_plus
from dotenv import load_dotenv
from sqlalchemy import create_engine

load_dotenv()  # reads variables from the .env file in the repo root

POSTGRES_USER = "postgres"
POSTGRES_PASSWORD = os.getenv("POSTGRES_PASSWORD")   # set this in your local .env file
POSTGRES_HOST = "localhost"
POSTGRES_PORT = "5432"
POSTGRES_DB = "analytics_course"

if not POSTGRES_PASSWORD:
    raise ValueError(
        "POSTGRES_PASSWORD not found. Create a .env file in the repo root "
        "with a line like: POSTGRES_PASSWORD=your_actual_password"
    )

engine = create_engine(
    f"postgresql://{POSTGRES_USER}:{quote_plus(POSTGRES_PASSWORD)}"
    f"@{POSTGRES_HOST}:{POSTGRES_PORT}/{POSTGRES_DB}"
)
pd.read_sql_query("SELECT version();", engine)
```

```python
query = """
SELECT * FROM chinook.track LIMIT 5;
"""
pd.read_sql_query(query, engine)
```

```python
query = """
SELECT category, sub_category FROM project_db.sales LIMIT 5;
"""
pd.read_sql_query(query, engine)
```

```python
query = """
SELECT ship_mode, category, sub_category, SUM(quantity) AS total_quantity
FROM project_db.sales
GROUP BY CUBE(ship_mode, category, sub_category)
LIMIT 30;
"""
pd.read_sql_query(query, engine)
```
